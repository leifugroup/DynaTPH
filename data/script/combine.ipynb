## Drop duplicates

```python
# # drop duplicates on raw db
pdbs = {}
for db in ['TRAIT',	'STCRDab', 'TCR3d', 'ATLAS', 'IEDB', 'VDJdb']:
    with open(f'{db}/0.pdb.txt', 'r') as f:
        pdbs[db] = f.read().split()

pdbs = set(itertools.chain.from_iterable(pdbs.values()))
with open("drop_duplicates/0.pdb.txt", 'w') as f:
    f.write('\n'.join([str(i) for i in pdbs]))
len(pdbs)
```
