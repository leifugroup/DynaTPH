```python
import json
import pandas as pd
pd.set_option('display.max_columns', None) 
pd.set_option('display.width', None)
pd.set_option('max_colwidth', 100)
```

```python
df = pd.read_csv("VDJdb/vdjdb-2025-02-21/vdjdb.txt", dtype=object, sep='\t')
print(f"VDJdb database has: {len(df)} records")
df.head()
```

```python
df['structure.id'] = df['meta'].apply(lambda s: json.loads(s)['structure.id'])

with open("VDJdb/0.pdb.txt", 'w') as f:
    f.write('\n'.join([str(i) for i in set(df['structure.id'].tolist())]))
```

## 1. Remove rows without PDB information

```python
df['structure.id'] = df['meta'].apply(lambda s: json.loads(s)['structure.id'])
df = df.loc[df['structure.id'].str.len()==4]
df['structure.id'] = df['structure.id'].str.lower()

print(f"VDJdb database has {len(df)} rows incloud PDB information")
df.head()
```

```python
with open("VDJdb/1.pdb.txt", 'w') as f:
    f.write('\n'.join([str(i) for i in set(df['structure.id'].tolist())]))
```

## 2. PDB drop duplicates

```python
df = df.drop_duplicates(subset=['structure.id'])
print(f"VDJdb database has: {len(df)} unique PDB")
df.head()
```

```python
with open("VDJdb/2.pdb.txt", 'w') as f:
    f.write('\n'.join([str(i) for i in set(df['structure.id'].tolist())]))
```

## 3. Keep human MHC

```python
print("mhc.a statistic")
df['mhc.a'].str[:3].value_counts()
```

```python
print("mhc.b statistic")
df['mhc.b'].str[:3].value_counts()
```

```python
df = df.loc[df['mhc.a'].str[:3]=='HLA']
print(f"VDJdb database has: {len(df)} unique human PDB")
df.head()
```

```python
df['structure.id'].rename('VDJdb').to_csv('VDJdb/VDJdb_pdb.csv', index=False)
```
