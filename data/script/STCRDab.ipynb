```python
df = pd.read_csv("STCRDab/STCRDab_raw.tsv", sep='\t', dtype=object)
print(f"STCRDab database has {len(df)} records has TCR structures")
df.head()
```

```python
with open("STCRDab/0.pdb.txt", 'w') as f:
    f.write('\n'.join([str(i) for i in set(df['pdb'].tolist())]))
```

## 1. Using rules to filter out pdb records

1. TCRtype = 'abTCR'
2. antigen_type = 'peptide'
3. mhc_type = 'MH1/MH2'

```python
df = df.loc[(df['TCRtype']=='abTCR') & (df['antigen_type']=='peptide') & df['mhc_type'].isin(['MH1','MH2'])] 
df['pdb'] = df['pdb'].str.lower()

print(f"STCRDab database has {len(df)} rows matched rules")
df.head()
```

```python
with open("STCRDab/1.pdb.txt", 'w') as f:
    f.write('\n'.join([str(i) for i in set(df['pdb'].tolist())]))
```

## 2. PDB drop duplicates

```python
df = df.drop_duplicates(subset=['pdb'])
print(f"STCRDab database has: {len(df)} unique PDB")
df.head()
```

```python
with open("STCRDab/2.pdb.txt", 'w') as f:
    f.write('\n'.join([str(i) for i in set(df['pdb'].tolist())]))
```

## 3. Select HLA

```python
df = df.loc[df['compound'].str.upper().str.contains('HLA')]
print(f"STCRDab database has: {len(df)} unique human PDB")
df.head()
```

```python
df['pdb'].rename('STCRDab').to_csv('STCRDab/STCRDab_pdb.csv', index=False)
```

```python
with open("STCRDab/3.pdb.txt", 'w') as f:
    f.write('\n'.join([str(i) for i in set(df['pdb'].tolist())]))
```

### extract human pdb from STCRDab webpage
link [https://opig.stats.ox.ac.uk/webapps/stcrdab-stcrpred/Browser?all=true]

```python
import requests
from bs4 import BeautifulSoup

response = requests.get("https://opig.stats.ox.ac.uk/webapps/stcrdab-stcrpred/Browser?all=true")

structures = BeautifulSoup(response.content, 'html.parser').find('table').findAll('tr')[1:]

pdb_df = pd.DataFrame([[cell.text for cell in row.findAll('td')[:6]] for row in structures], columns=['PDB', 'Species', 'Method', 'Resolution', 'Affinity', 'TCR_Chains'])

pdb_df.to_csv("STCRDab/pdb.csv", index=False)
print(f"pdb types: {len(pdb_df)}")
```
