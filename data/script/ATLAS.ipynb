```python
import pandas as pd
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
pd.set_option('max_colwidth', 100)
```

```python
# !mkdir ATLAS
# !wget https://atlas.wenglab.org/web/tables/ATLAS.xlsx -O ATLAS/ATLAS.xlsx
```

```python
df = pd.read_excel("ATLAS/ATLAS.xlsx", dtype=object).drop(columns=['Index'])
print(f"ATLAS database has {len(df)} records has records")
df.head()
```

```python
with open("ATLAS/0.pdb.txt", 'w') as f:
    f.write('\n'.join([str(i) for i in set(df['true_PDB'].tolist())]))
```

## 1. Remove rows without PDB information

```python
df = df.dropna(subset=['true_PDB'])
df['true_PDB'] = df['true_PDB'].str.lower()

print(f"ATLAS database has {len(df)} rows incloud PDB information")
df.head()
```

```python
with open("ATLAS/1.pdb.txt", 'w') as f:
    f.write('\n'.join([str(i) for i in set(df['true_PDB'].tolist())]))
```

## 2. PDB drop duplicates

```python
df = df.drop_duplicates(subset=['true_PDB'])
print(f"ATLAS database has: {len(df)} unique PDB")
df.head()
```

```python
with open("ATLAS/2.pdb.txt", 'w') as f:
    f.write('\n'.join([str(i) for i in set(df['true_PDB'].tolist())]))
```

## 3. Keep human MHC

```python
df = df.loc[df['MHCname'].str.upper().str[:3]=='HLA']
print(f"ATLAS database has: {len(df)} unique human PDB")
df.head()
```

```python
df['true_PDB'].rename('ATLAS').to_csv('ATLAS/ATLAS_pdb.csv', index=False)
```

```python
with open("ATLAS/3.pdb.txt", 'w') as f:
    f.write('\n'.join([str(i) for i in set(df['true_PDB'].tolist())]))
```
