```python
import pandas as pd
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
pd.set_option('max_colwidth', 100)
```

```python
# !mkdir TCR3d
# !wget https://tcr3d.ibbr.umd.edu/static/download/tcr_complexes_data.tsv -O TCR3d/TCR3d_raw.tsv
```

```python
df = pd.read_csv("TCR3d/TCR3d_raw.tsv", sep='\t', dtype=object)
print(f"TCR3d database has {len(df)} records has TCR-peptide-MHC complex structures")
df.head()
```

```python
with open("TCR3d/0.pdb.txt", 'w') as f:
    f.write('\n'.join([str(i) for i in set(df['PDB_ID'].tolist())]))
```

## 1. Fill empty values in MHC_allele

```python
df.loc[df['PDB_ID'].isin(['8rj5', '8yiv', '8yj2', '9c3e']), 'MHC_allele'] = 'HLA'
df['PDB_ID'] = df['PDB_ID'].str.lower()

print(f"TCR3d database has {len(df)} rows matched rules")
df.loc[df['PDB_ID'].isin(['8rj5', '8yiv', '8yj2', '9c3e'])]
```

```python
with open("TCR3d/1.pdb.txt", 'w') as f:
    f.write('\n'.join([str(i) for i in set(df['PDB_ID'].tolist())]))
```

## 2. PDB drop duplicates

```python
df = df.drop_duplicates(subset=['PDB_ID'])
print(f"TCR3d database has: {len(df)} unique PDB")
df.head()
```

```python
with open("TCR3d/2.pdb.txt", 'w') as f:
    f.write('\n'.join([str(i) for i in set(df['PDB_ID'].tolist())]))
```

## 3. Select HLA

```python
df = df.loc[df['MHC_allele'].str.upper().str[:3]=='HLA']
print(f"TCR3d database has: {len(df)} unique human PDB")
df.head()
```

```python
df['PDB_ID'].rename('TCR3d').to_csv('TCR3d/TCR3d_pdb.csv', index=False)
```

```python
with open("TCR3d/3.pdb.txt", 'w') as f:
    f.write('\n'.join([str(i) for i in set(df['PDB_ID'].tolist())]))
```
