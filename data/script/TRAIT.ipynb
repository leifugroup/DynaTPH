```python
df = pd.read_excel("TRAIT/Interactive_TCR-pMHC_Pairs.xlsx", dtype=object)
df.to_csv('TRAIT/TRAIT_raw.csv', index=False)
print(f"TRAIT database has: {len(df)} records")
df.head()
```

```python
with open("TRAIT/0.pdb.txt", 'w') as f:
    f.write('\n'.join([str(i) for i in set(df['Structure'].tolist())]))
```

## 1. Remove rows without PDB information

```python
df = df.dropna(subset=['Structure'])
df['Structure'] = df['Structure'].str.lower()

print(f"TRAIT database has {len(df)} rows incloud PDB information")
df.head()
```

```python
with open("TRAIT/1.pdb.txt", 'w') as f:
    f.write('\n'.join([str(i) for i in set(df['Structure'].tolist())]))
```

## 2. PDB drop duplicates

```python
df = df.drop_duplicates(subset=['Structure'])
print(f"TRAIT database has: {len(df)} unique PDB")
df.head()
```

```python
with open("TRAIT/2.pdb.txt", 'w') as f:
    f.write('\n'.join([str(i) for i in set(df['Structure'].tolist())]))
```

## 3. Keep human MHC

```python
print("MHC_A statistic")
df['MHC_A'].str[:3].value_counts()
```

```python
print("MHC_B statistic")
df['MHC_B'].str[:3].value_counts()
```

```python
df = df.loc[df['MHC_A'].str[:3]=='HLA']
print(f"TRAIT database has: {len(df)} unique human PDB")
df.head()
```
