```python
df = pd.read_csv("IEDB/tcell_full_v3.csv", header=[0,1], dtype=object)
print(f"IEDB database has {len(df)} records has records")
df.head()
```

```python
with open("IEDB/0.pdb.txt", 'w') as f:
    f.write('\n'.join([str(i) for i in set(df[('Complex', 'PDB ID')].tolist())]))
```

## 1. PDB Filter

```python
df = df.dropna(subset=[('Complex', 'PDB ID')])
df = df.loc[(df[('Epitope', 'Object Type')] == 'Linear peptide') & (df[('Assay', 'Qualitative Measurement')] == 'Positive')]
df = df.loc[df[('Complex', 'PDB ID')].str.len()==4]
df[('Complex', 'PDB ID')] = df[('Complex', 'PDB ID')].str.lower()

print(f"IEDB database has {len(df)} rows incloud PDB information")
df.head()
```

```python
with open("IEDB/1.pdb.txt", 'w') as f:
    f.write('\n'.join([str(i) for i in set(df[('Complex', 'PDB ID')].tolist())]))
```

## 2. PDB drop duplicates

```python
df = df.drop_duplicates(subset=[('Complex', 'PDB ID')])
print(f"IEDB database has: {len(df)} unique PDB")
df.head()
```

```python
with open("IEDB/2.pdb.txt", 'w') as f:
    f.write('\n'.join([str(i) for i in set(df[('Complex', 'PDB ID')].tolist())]))
```

## 3. Keep human MHC

```python
df = df.loc[df[('MHC Restriction', 'Name')].str.upper().str[:3]=='HLA']
print(f"ATLAS database has: {len(df)} unique human PDB")
df.head()
```

```python
df[('Complex', 'PDB ID')].rename('IEDB').to_csv('IEDB/IEDB_pdb.csv', index=False)
```

```python
with open("IEDB/3.pdb.txt", 'w') as f:
    f.write('\n'.join([str(i) for i in set(df[('Complex', 'PDB ID')].tolist())]))
```
